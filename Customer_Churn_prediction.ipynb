Feature Engineering

```python
df["Signup_Date"] = pd.to_datetime(
    df["Signup_Date"],
    errors="coerce"
)

df["Signup_Year"] = df["Signup_Date"].dt.year
df["Signup_Month"] = df["Signup_Date"].dt.month
df["Signup_Day"] = df["Signup_Date"].dt.day
df["Signup_DayOfWeek"] = df["Signup_Date"].dt.dayofweek
```

```python
df["Signup_Is_Weekend"] = (
    df["Signup_DayOfWeek"] >= 5
).astype(int)
```

```python
df["Tenure_Years"] = df["Tenure_Months"] / 12
```

```python
df["Total_Usage_Time"] = (
    df["Usage_Frequency"] *
    df["Avg_Session_Duration_Min"]
)
```

```python
df["Total_Support_Issues"] = (
    df["Support_Tickets_Raised"] +
    df["Complaint_Count"]
)
```

```python
df["Has_Late_Payment"] = (
    df["Late_Payment_Count"] > 0
).astype(int)
```

```python
df["Has_Referral"] = (
    df["Referral_Count"] > 0
).astype(int)
```

```python
df["Avg_Monthly_Charge"] = (
    df["Total_Charges"] /
    df["Tenure_Months"].replace(0, np.nan)
)
```

```python
support_threshold = df["Support_Tickets_Raised"].median()

df["High_Support_Usage"] = (
    df["Support_Tickets_Raised"] >
    support_threshold
).astype(int)
```

```python
complaint_threshold = df["Complaint_Count"].median()

df["High_Complaint"] = (
    df["Complaint_Count"] >
    complaint_threshold
).astype(int)
```

```python
df["Engagement_Level"] = pd.cut(
    df["Usage_Frequency"],
    bins=[-np.inf, 10, 30, np.inf],
    labels=["Low", "Medium", "High"]
)
```

```python
# ============================================
# LOGISTIC REGRESSION - FULL MODEL METRICS
# ============================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    log_loss,
    brier_score_loss
)


# ============================================
# 1. PREDICTIONS
# ============================================

# Class prediction: 0 or 1
y_pred = logistic_model.predict(X_test_scaled)

# Probability prediction
y_prob = logistic_model.predict_proba(X_test_scaled)[:, 1]


# ============================================
# 2. ACCURACY
# ============================================

accuracy = accuracy_score(y_test, y_pred)


# ============================================
# 3. PRECISION
# ============================================

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)


# ============================================
# 4. RECALL
# ============================================

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)


# ============================================
# 5. F1-SCORE
# ============================================

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)


# ============================================
# 6. ROC-AUC
# ============================================

roc_auc = roc_auc_score(
    y_test,
    y_prob
)


# ============================================
# 7. PR-AUC
# ============================================

pr_auc = average_precision_score(
    y_test,
    y_prob
)


# ============================================
# 8. CONFUSION MATRIX
# ============================================

cm = confusion_matrix(
    y_test,
    y_pred
)


# ============================================
# 9. LOG LOSS / CROSS-ENTROPY
# ============================================

logloss = log_loss(
    y_test,
    y_prob
)


# ============================================
# 10. BRIER SCORE
# ============================================

brier = brier_score_loss(
    y_test,
    y_prob
)


# ============================================
# 11. PRINT ALL METRICS
# ============================================

print("=" * 60)
print("       LOGISTIC REGRESSION - MODEL METRICS")
print("=" * 60)

print(f"Accuracy     : {accuracy:.4f}  ({accuracy*100:.2f}%)")
print(f"Precision    : {precision:.4f}  ({precision*100:.2f}%)")
print(f"Recall       : {recall:.4f}  ({recall*100:.2f}%)")
print(f"F1-Score     : {f1:.4f}  ({f1*100:.2f}%)")
print(f"ROC-AUC      : {roc_auc:.4f}")
print(f"PR-AUC       : {pr_auc:.4f}")
print(f"Log Loss     : {logloss:.4f}")
print(f"Brier Score  : {brier:.4f}")

print("\nConfusion Matrix:")
print(cm)

print("=" * 60)


# ============================================
# 12. CONFUSION MATRIX VISUALIZATION
# ============================================

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Not Churn", "Churn"],
    yticklabels=["Not Churn", "Churn"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Logistic Regression - Confusion Matrix")

plt.show()
```

```python
# ============================================
# LIFT & GAIN DATA
# ============================================

lift_df = pd.DataFrame({
    "Actual": y_test.values,
    "Probability": y_prob
})

# Sort from highest churn probability to lowest
lift_df = lift_df.sort_values(
    by="Probability",
    ascending=False
).reset_index(drop=True)

# Cumulative actual churners
lift_df["Cumulative_Churn"] = (
    lift_df["Actual"].cumsum()
)

# Total actual churners
total_churners = lift_df["Actual"].sum()

# Cumulative Gain
lift_df["Cumulative_Gain"] = (
    lift_df["Cumulative_Churn"] /
    total_churners
)

# Percentage of customers targeted
lift_df["Customer_Percentage"] = (
    np.arange(1, len(lift_df) + 1) /
    len(lift_df)
)

# Cumulative Lift
lift_df["Cumulative_Lift"] = (
    lift_df["Cumulative_Gain"] /
    lift_df["Customer_Percentage"]
)
```

```python
plt.figure(figsize=(8, 6))

plt.plot(
    lift_df["Customer_Percentage"] * 100,
    lift_df["Cumulative_Lift"]
)

plt.axhline(
    y=1,
    linestyle="--"
)

plt.xlabel("Customers Targeted (%)")
plt.ylabel("Cumulative Lift")
plt.title("Lift Chart - Logistic Regression")

plt.grid()

plt.show()
```

```python
# ============================================
# COST-SENSITIVE METRICS
# ============================================

FP_COST = 100    # Cost of incorrectly targeting a customer
FN_COST = 1000   # Cost of missing an actual churn customer

tn, fp, fn, tp = cm.ravel()

false_positive_cost = fp * FP_COST
false_negative_cost = fn * FN_COST

total_cost = (
    false_positive_cost +
    false_negative_cost
)

cost_per_customer = (
    total_cost / len(y_test)
)

print("\n" + "=" * 60)
print("       COST-SENSITIVE METRICS")
print("=" * 60)

print(f"True Negatives      : {tn}")
print(f"False Positives     : {fp}")
print(f"False Negatives     : {fn}")
print(f"True Positives      : {tp}")

print(f"\nFalse Positive Cost : ₹{false_positive_cost}")
print(f"False Negative Cost : ₹{false_negative_cost}")
print(f"Total Business Cost : ₹{total_cost}")
print(f"Cost Per Customer   : ₹{cost_per_customer:.2f}")

print("=" * 60)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create Lift DataFrame
lift_df = pd.DataFrame({
    "Actual": y_test.values,
    "Probability": y_prob_xgb
})

# Sort customers by predicted churn probability
lift_df = lift_df.sort_values(
    by="Probability",
    ascending=False
).reset_index(drop=True)

# Customer rank
lift_df["Customer_Rank"] = np.arange(
    1,
    len(lift_df) + 1
)

# Cumulative churners
lift_df["Cumulative_Churn"] = (
    lift_df["Actual"].cumsum()
)

# Total actual churners
total_churners = lift_df["Actual"].sum()

# Cumulative gain
lift_df["Cumulative_Gain"] = (
    lift_df["Cumulative_Churn"] /
    total_churners
)

# Percentage of customers contacted
lift_df["Customer_Percentage"] = (
    lift_df["Customer_Rank"] /
    len(lift_df)
)

# Lift
lift_df["Lift"] = (
    lift_df["Cumulative_Gain"] /
    lift_df["Customer_Percentage"]
)

print(lift_df.head(10))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create dataframe
lift_gain_df = pd.DataFrame({
    "Actual": y_test.values,
    "Probability": y_prob_xgb_tuned
})

# Sort customers by predicted churn probability
lift_gain_df = lift_gain_df.sort_values(
    by="Probability",
    ascending=False
).reset_index(drop=True)

# Cumulative churners
lift_gain_df["Cumulative_Churn"] = (
    lift_gain_df["Actual"].cumsum()
)

# Total actual churners
total_churners = lift_gain_df["Actual"].sum()

# Cumulative Gain
lift_gain_df["Cumulative_Gain"] = (
    lift_gain_df["Cumulative_Churn"] / total_churners
)

# Percentage of customers targeted
lift_gain_df["Customer_Percentage"] = (
    np.arange(1, len(lift_gain_df) + 1)
    / len(lift_gain_df)
)

# Cumulative Lift
lift_gain_df["Cumulative_Lift"] = (
    lift_gain_df["Cumulative_Gain"]
    / lift_gain_df["Customer_Percentage"]
)

print(lift_gain_df.head())
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    lift_gain_df["Customer_Percentage"] * 100,
    lift_gain_df["Cumulative_Gain"] * 100,
    label="Tuned XGBoost"
)

# Random baseline
plt.plot(
    [0, 100],
    [0, 100],
    linestyle="--",
    label="Random Baseline"
)

plt.xlabel("Percentage of Customers Targeted (%)")
plt.ylabel("Cumulative Gain (%)")
plt.title("Tuned XGBoost - Cumulative Gain Chart")

plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    lift_gain_df["Customer_Percentage"] * 100,
    lift_gain_df["Cumulative_Lift"],
    label="Tuned XGBoost"
)

# Random baseline = Lift 1
plt.axhline(
    y=1,
    linestyle="--",
    label="Random Baseline"
)

plt.xlabel("Percentage of Customers Targeted (%)")
plt.ylabel("Cumulative Lift")
plt.title("Tuned XGBoost - Lift Chart")

plt.legend()
plt.grid(True)

plt.show()
```

```python
percentages = [10, 20, 30, 40, 50]

print("===== Lift & Gain Summary =====")

for percentage in percentages:

    index = int(
        len(lift_gain_df) * percentage / 100
    ) - 1

    gain = (
        lift_gain_df.loc[index, "Cumulative_Gain"]
        * 100
    )

    lift = lift_gain_df.loc[
        index, "Cumulative_Lift"
    ]

    print(
        f"Top {percentage}% Customers "
        f"-> Gain: {gain:.2f}% "
        f"| Lift: {lift:.2f}"
    )
```
